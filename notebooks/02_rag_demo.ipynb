1. Store the company's information in a vector database (ChromaDB)
2. When generating a simulation, retrieve the relevant info
3. Inject it into the prompt as context

## 1. What Are Embeddings?

An **embedding** is a way to represent text as a list of numbers (a vector).
Similar texts produce similar vectors â€” this is the key insight.

For example:
- `"software engineer"` â†’ `[0.21, -0.54, 0.88, ...]`
- `"developer role"` â†’ `[0.19, -0.51, 0.85, ...]`  â† very similar!
- `"marketing intern"` â†’ `[-0.43, 0.12, -0.22, ...]` â† very different

A **vector database** (like ChromaDB) stores these vectors and lets you ask:
> "Find me the texts most similar to this query"

This is called a **semantic search** â€” it understands *meaning*, not just keywords.

## 2. Setting Up ChromaDB

```python
import chromadb
from chromadb.utils import embedding_functions

# In-memory client for this notebook (no files written to disk)
client = chromadb.Client()

# Default embedding function (uses a small local model â€” no API key needed)
ef = embedding_functions.DefaultEmbeddingFunction()

# Create a collection â€” think of it as a table in a regular database
collection = client.create_collection(
    name='company_contexts',
    embedding_function=ef
)

print('ChromaDB collection created:', collection.name)
```

## 3. Storing Company Information

```python
# Let's create 3 fictional companies with different profiles
companies = [
    {
        'id': 'company_001',
        'name': 'DataFlow Analytics',
        'text': (
            'DataFlow Analytics is a B2B SaaS company specializing in real-time data pipelines. '
            'The engineering team uses Python, Kafka, dbt, and Snowflake. '
            'Interns work on data transformation scripts, dashboard development, and ETL pipeline monitoring. '
            'The team values clean code, documentation, and data accuracy.'
        )
    },
    {
        'id': 'company_002',
        'name': 'GreenLeaf E-commerce',
        'text': (
            'GreenLeaf is a sustainable e-commerce platform selling eco-friendly products. '
            'The marketing team uses HubSpot, Google Analytics, Canva, and Shopify. '
            'Interns help with content creation, social media campaigns, email marketing, and SEO analysis. '
            'The team values creativity, sustainability mindset, and data-driven decisions.'
        )
    },
    {
        'id': 'company_003',
        'name': 'FinEdge Capital',
        'text': (
            'FinEdge Capital is an early-stage fintech startup building investment tools for retail investors. '
            'The product team uses Figma, Notion, Mixpanel, and Jira. '
            'Interns conduct user research, create wireframes, write PRDs, and analyze user behavior data. '
            'The team values user empathy, clear communication, and fast iteration.'
        )
    }
]

# Add all companies to ChromaDB
collection.add(
    ids=[c['id'] for c in companies],
    documents=[c['text'] for c in companies],
    metadatas=[{'name': c['name']} for c in companies]
)

print(f'Stored {collection.count()} companies in ChromaDB')
```

## 4. Semantic Search â€” Finding the Right Company

```python
# Try different queries and see which company matches
queries = [
    'Python data engineering pipeline',
    'social media content marketing',
    'product design user research'
]

for query in queries:
    result = collection.query(query_texts=[query], n_results=1)
    matched_company = result['metadatas'][0][0]['name']
    matched_text = result['documents'][0][0][:80] + '...'
    print(f'Query: "{query}"')
    print(f'  Best match: {matched_company}')
    print(f'  Snippet: {matched_text}')
    print()
```

**Observation:** ChromaDB matched each query to the correct company purely based on semantic similarity â€” no keyword rules, just meaning.


## 5. RAG in Action â€” Context-Aware Simulation Generation

```python
from groq import Groq
import json

groq_client = Groq(api_key=GROQ_API_KEY)

def generate_simulation_with_rag(position: str, company_query: str) -> dict:
    """
    Generate a simulation using RAG:
    1. Find the most relevant company context from ChromaDB
    2. Inject it into the prompt
    3. Let the LLM use it to generate a tailored simulation
    """
    # Step 1: Retrieve relevant company context
    result = collection.query(query_texts=[company_query], n_results=1)
    company_context = result['documents'][0][0]
    company_name = result['metadatas'][0][0]['name']
    
    print(f'Retrieved context for: {company_name}')
    print(f'Context snippet: {company_context[:100]}...')
    print()
    
    # Step 2: Build prompt with the retrieved context injected
    prompt = f"""You are designing an internship simulation task.

COMPANY CONTEXT (use this to make the task specific and realistic):
{company_context}

POSITION: {position}

Create ONE realistic task for this intern. Output ONLY valid JSON:
{{
  "title": "task title",
  "scenario": "the situation the intern is placed in",
  "task_description": "what they must do",
  "tools_to_use": ["tool1", "tool2"],
  "expected_output": "what a good response looks like",
  "time_limit_minutes": 20
}}"""
    
    # Step 3: Call LLM with the context-enriched prompt
    response = groq_client.chat.completions.create(
        model='llama-3.3-70b-versatile',
        temperature=0.7,
        messages=[
            {'role': 'system', 'content': 'You output ONLY valid JSON. No markdown, no explanation.'},
            {'role': 'user', 'content': prompt}
        ]
    )
    
    return json.loads(response.choices[0].message.content)


# Generate a simulation for a data engineering intern
simulation = generate_simulation_with_rag(
    position='Data Engineering Intern',
    company_query='Python data pipelines ETL'
)

print('Generated Simulation:')
print(json.dumps(simulation, indent=2))
```

## 6. RAG vs. No-RAG â€” Side-by-Side Comparison

This is the most important cell. See how much difference context makes.

```python
def generate_simulation_no_rag(position: str) -> dict:
    """Generate a simulation WITHOUT any company context â€” generic prompt only."""
    response = groq_client.chat.completions.create(
        model='llama-3.3-70b-versatile',
        temperature=0.7,
        messages=[
            {'role': 'system', 'content': 'You output ONLY valid JSON. No markdown.'},
            {'role': 'user', 'content': f"""Create ONE internship simulation task for a {position}.
Output JSON: {{"title": "", "scenario": "", "task_description": "", "tools_to_use": [], "expected_output": "", "time_limit_minutes": 20}}"""}
        ]
    )
    return json.loads(response.choices[0].message.content)


print('=' * 60)
print('WITHOUT RAG (generic):')
print('=' * 60)
no_rag = generate_simulation_no_rag('Data Engineering Intern')
print(json.dumps(no_rag, indent=2))

print()
print('=' * 60)
print('WITH RAG (DataFlow Analytics specific):')
print('=' * 60)
with_rag = generate_simulation_with_rag(
    position='Data Engineering Intern',
    company_query='Python data pipelines ETL'
)
print(json.dumps(with_rag, indent=2))
```

## Answers to Week 2 Questions

### 1. What would happen if two very similar companies were stored? Would retrieval still work correctly?

If two companies are very similar, retrieval may not always select the company we expect. Since semantic search works based on similarity between embeddings, two companies with similar contexts could have very close representations, and the retrieval system might return the wrong one. We actually observed a similar issue in this notebook when the product management query retrieved GreenLeaf E-commerce instead of FinEdge Capital.

### 2. Why do we use `temperature=0.7` for generation but would use `0.2` for evaluation?

During simulation generation, the model acts as a creator, so we want to allow some variety and creativity instead of producing almost identical tasks every time. This is why a temperature such as 0.7 is useful. During evaluation, however, the model acts more like a judge. Consistency is more important than creativity because similar candidate responses should receive similar evaluations. Therefore, a lower temperature such as 0.2 is more appropriate.

### 3. How would you handle a company that submits very little information â€” what's the minimum context needed?

The minimum context should contain enough information for the model to fill the important parts of the simulation output in a company-specific way. For example, it should at least include what the company or team does, what kind of work the intern or position is responsible for, and the main tools or technologies they use. If this information is missing, the model can still generate a valid JSON output, but it would have to fill the missing details using more generic information, making the simulation less company-specific.
