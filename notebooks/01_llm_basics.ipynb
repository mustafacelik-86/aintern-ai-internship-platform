## 1. What IS a Large Language Model?

An LLM (like Llama 3.1, GPT-4, Claude) is a neural network trained on massive amounts of text.

It doesn't 'understand' text the way humans do. It predicts the **most likely next token (word piece)** given everything that came before.

That's it. Everything you see â€” code generation, reasoning, creative writing â€” comes from this one mechanism.

### Key concepts:

| Concept | What it means |
|---------|---------------|
| **Token** | ~0.75 words. Models think in tokens, not words. |
| **Context window** | Max tokens the model can 'see' at once (like short-term memory) |
| **Temperature** | Randomness. 0 = deterministic, 1 = creative/unpredictable |
| **System prompt** | Instructions given to the model before the conversation starts |
| **Prompt** | The actual message you send |
| **Completion** | The model's response |

## 2. Your First LLM Call

````python
from groq import Groq

client = Groq(api_key=GROQ_API_KEY)

# This is the simplest possible LLM call
response = client.chat.completions.create(
    model='llama-3.1-8b-instant',  # Faster/cheaper for experiments
    messages=[
        {'role': 'user', 'content': 'What is an internship simulation?'}
    ]
)

print(response.choices[0].message.content)
````

## 3. System Prompts Change Everything

A **system prompt** tells the model who it is and how it should behave.
Watch what happens when we add one:

````python
# Same question â€” but with a system prompt
response = client.chat.completions.create(
    model='llama-3.1-8b-instant',
    messages=[
        {
            'role': 'system',
            'content': 'You are a concise, no-nonsense hiring expert. '
                       'Answer in bullet points only. Max 3 bullets.'
        },
        {
            'role': 'user',
            'content': 'What is an internship simulation?'
        }
    ]
)

print(response.choices[0].message.content)
````

## 4. Temperature: Creativity vs. Consistency

Run this cell multiple times and see how the output changes:

````python
# Try: temperature=0.0, then temperature=0.9, then temperature=1.5
TEMPERATURE = 1.5

response = client.chat.completions.create(
    model='llama-3.1-8b-instant',
    temperature=TEMPERATURE,
    messages=[
        {
            'role': 'user',
            'content': 'Give me ONE creative task idea for a marketing intern simulation. '
                       'Just the task title and one sentence description.'
        }
    ]
)

print(f'Temperature: {TEMPERATURE}')
print(response.choices[0].message.content)
````

**Observation:** At temperature=0 you get the same answer every time. At 0.9 it varies. 
In AIntern we use ~0.75 for simulation generation (we WANT variety) and ~0.2 for evaluation (we WANT consistency).

### Temperature Observation

- At `temperature = 0.0`, the model produced the same and consistent response in repeated runs.
- At `temperature = 0.9`, the responses became more varied and creative, but less closely related to each other.
- At `temperature = 1.5`, consistency decreased further and the outputs became much more unpredictable.

## 5. Getting Structured (JSON) Output

The biggest challenge with LLMs: they're verbose. For our pipeline we need clean JSON.
Here's how we force it:

````python
import json

response = client.chat.completions.create(
    model='llama-3.1-8b-instant',
    temperature=0.3,
    messages=[
        {
            'role': 'system',
            'content': 'You output ONLY valid JSON. No markdown, no explanation.'
        },
        {
            'role': 'user',
            'content': '''
Create ONE internship task for a software engineering intern at a startup.

Output format:
{
  "title": "task title",
  "description": "what the candidate must do",
  "expected_skills": ["skill1", "skill2"],
  "time_limit_minutes": 15
}
'''
        }
    ]
)

raw_text = response.choices[0].message.content
print('Raw response:')
print(raw_text)
print()

# Parse it
task = json.loads(raw_text)
print('Parsed successfully âœ…')
print(f'Task title: {task["title"]}')
print(f'Skills: {task["expected_skills"]}')
````

## 6. LangChain: Chains > Raw API Calls

LangChain lets us build **pipelines** (chains) instead of calling the API manually each time.
This is what the real AIntern code uses:

````python
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import JsonOutputParser

# 1. Define the LLM
llm = ChatGroq(
    api_key=GROQ_API_KEY,
    model='llama-3.1-8b-instant',
    temperature=0.3
)

# 2. Define the prompt template (with variables)
prompt = ChatPromptTemplate.from_messages([
    ('system', 'You output ONLY valid JSON. No markdown, no explanation.'),
    ('human', '''
Create ONE internship simulation task for the following position.

Company: {company}
Position: {position}

Output: {{"title": "...", "description": "...", "expected_skills": [...], "time_limit_minutes": 15}}
''')
])

# 3. Build the chain: prompt â†’ llm â†’ parse JSON
chain = prompt | llm | JsonOutputParser()

# 4. Run it with variables
result = chain.invoke({
    'company': 'Acme SaaS',
    'position': 'Product Marketing Intern'
})

print('Result type:', type(result))  # Should be dict, not string!
print('Task:', result)
````
